#Rock Paper Scissors game
สร้างเกม เป่า ยิ้ง ฉุบ

* ฝ่ายใดชนะ +1 คะแนน
* เสมอกัน ไม่เพิ่มคะแนน
* ถ้า user บอกว่า exit ให้สรุปคะแนน

```python
import random

def rps_game():
    # กำหนดค่า action ให้คอมเอาไปสุ่ม
    actions = ["r", "p", "s"]

    player_win = 0
    com_win = 0
    tie = 0

    print("Welcome to Rock-Paper-Scissors!")

    while True:
        # ใส่คำอธิบายให้ผู้เล่นอ่านและเลือกตัวอักษรแทนท่าทางของมือ
        # สร้าง player_move ไว้ใส่ตัวเลือกของผู้เล่น
        player_move = input("\nChoose your move: rock[r], paper[p], scissors[s], exit[x]: ")

        # check ว่าผู้เล่นตอบมาตรงตามที่กำหนดไว้หรือไม่ ถ้าไม่ ให้พิมพ์ใหม่
        if player_move not in ["r", "p", "s", "x"]:
            print("Invalid choice. Please choose r, p, s, or x to exit.")
            continue


        # สร้าง computer_move ใส่ค่าที่คอมสุ่มมา
        computer_move = random.choice(actions)

        # ถ้าผู้เล่นเลือก x ให้แสดงคะแนนสะสม แล้วออกจากเกม
        if player_move == "x":
            print("\nGoodbye! :D")
            print(f"Final Score -> You: {player_win} | Computer: {com_win}")
            break


        print(f"\nPlayer Move: {player_move}")
        print(f"Computer Move: {computer_move}")

        # กำหนดเงื่อนไขแพ้ชนะและเสมอ
        if player_move == computer_move:
            print(">>> TIE!")
            tie += 1
        elif (player_move == "r" and computer_move == "s") or \
             (player_move == "s" and computer_move == "p") or \
             (player_move == "p" and computer_move == "r"):
            print(">>> YOU WIN!")
            player_win += 1
        else:
            print(">>> YOU LOSE!")
            com_win += 1

        # แสดงผลรวมคะแนน
        print(f"SCORE - WIN: {player_win} | LOSS: {com_win} | TIE: {tie}")

```

ลองเล่นเกมดู

```python
rps_game()
```
