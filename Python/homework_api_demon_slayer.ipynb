##for loop

ดึงข้อมูลทีละหลายคนต้องใช้ for loop

index เริ่มจาก 0 เลข id จึงต้องเป็น i+1

```python
import requests
import time

characters = []

for i in range(45):
    url = f"https://www.demonslayer-api.com/api/v1/characters?id={i+1}"
    try:
        response = requests.get(url)
        response_json = response.json()
        data = [response_json['content'][0]['name'],
                response_json['content'][0]['age'],
                response_json['content'][0]['affiliation']['name'],
                response_json['content'][0]['combat_style'][0]['name']]
        characters.append(data)
        print(f"Success: {i+1}")
        time.sleep(1)
    except:
        print("No character data found.")
```

แก้ต่อโดยเพิ่ม if else และใช้ .get method

```python
import requests
import time

characters = []

for i in range(45):
    url = f"https://www.demonslayer-api.com/api/v1/characters?id={i+1}"

    try:
        response = requests.get(url)
        response_json = response.json()
        content = response_json.get('content', [])
        char = content[0]

        combat_styles = char.get('combat_style', [])
        combat_name = combat_styles[0].get('name', 'None') if combat_styles else "None"

        data = [char.get('name', 'Unknown'),
                char.get('age', 'Unknown'),
                char.get('gender', 'Unknown'),
                char.get('race', 'Unknown'),
                char.get('affiliation', {}).get('name', 'None'),
                combat_name]

        characters.append(data)
        print(f"Success: {i+1}")

    except:
        print(f"Error on ID {i+1}")

    time.sleep(1)

```

```python
print(characters)
```

```python
import pandas as pd

demonslayer = pd.DataFrame(characters,
              columns = ["name",
                         "age",
                         "gender",
                         "race",
                         "affiliation",
                         "combat style"])
```
